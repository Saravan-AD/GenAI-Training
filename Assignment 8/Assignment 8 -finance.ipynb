```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END
from langchain_openai import ChatOpenAI
import os
```

```python
os.environ["OPENROUTER_API_KEY"]="api_key"

llm = ChatOpenAI(
    model="openai/gpt-4o-mini",
    api_key=os.environ["OPENROUTER_API_KEY"],
    base_url="https://openrouter.ai/api/v1"
)
```

```python
class FinanceState(TypedDict):

    applicant_details: str
    salary_slips: str
    bank_statements: str
    loan_amount: str
    employment_type: str
    credit_score: str

    applicant_profile: str
    income_stability: str
    missing_documents: str
    spending_patterns: str
    financial_behavior: str
    risk_assessment: str

    credit_validation: str
    fraud_detection: str
    kyc_verification: str
    employment_verification: str
    debt_to_income_ratio: str

    next_route: str
    assigned_path: str

    risk_summary: str
    approval_rationale: str
    loan_recommendation: str
    interest_slab: str
    applicant_notification: str

    # FINAL
    final_decision_report: str
```

```python
def extract_profile(state: FinanceState):

    prompt = f"""
    You are a Financial Applicant Profile Extraction Agent.

    Extract:
    1. Employment type
    2. Salary details
    3. Loan request amount
    4. Applicant category
    5. Existing liabilities
    6. Financial obligations

    Applicant Details:
    {state['applicant_details']}

    Return structured applicant profile.
    """

    response = llm.invoke(prompt)

    return {
        "applicant_profile": response.content
    }
```

```python
def analyze_income_stability(state: FinanceState):

    prompt = f"""
    You are an Income Stability Analysis Agent.

    Analyze:
    1. Salary consistency
    2. Income stability
    3. Employment continuity
    4. Financial reliability
    5. Income risk indicators

    Salary Slips:
    {state['salary_slips']}

    Bank Statements:
    {state['bank_statements']}
    """

    response = llm.invoke(prompt)

    return {
        "income_stability": response.content
    }
```

```python
def check_missing_documents(state: FinanceState):

    prompt = f"""
    You are a Financial Document Validation Agent.

    Check for missing:
    - Salary slips
    - Bank statements
    - Tax documents
    - Identity proof
    - Employment proof

    Applicant Details:
    {state['applicant_details']}
    """

    response = llm.invoke(prompt)

    return {
        "missing_documents": response.content
    }
```

```python
def analyze_spending_patterns(state: FinanceState):

    prompt = f"""
    You are a Spending Pattern Analysis Agent.

    Analyze:
    1. Monthly spending
    2. Savings behavior
    3. Expense-to-income ratio
    4. Financial discipline
    5. Luxury spending patterns

    Bank Statements:
    {state['bank_statements']}
    """

    response = llm.invoke(prompt)

    return {
        "spending_patterns": response.content
    }

```

```python
def analyze_financial_behavior(state: FinanceState):

    prompt = f"""
    You are a Financial Behavior Assessment Agent.

    Analyze:
    1. Repayment discipline
    2. Existing debts
    3. Credit behavior
    4. Financial responsibility
    5. Risk indicators

    Applicant Details:
    {state['applicant_details']}
    """

    response = llm.invoke(prompt)

    return {
        "financial_behavior": response.content
    }
```

```python
def assess_risk(state: FinanceState):

    prompt = f"""
    You are a Loan Risk Assessment Agent.

    Analyze:

    Applicant Profile:
    {state['applicant_profile']}

    Income Stability:
    {state['income_stability']}

    Spending Patterns:
    {state['spending_patterns']}

    Financial Behavior:
    {state['financial_behavior']}

    Determine:
    - Low Risk
    - Medium Risk
    - High Risk
    - Critical Risk

    Provide detailed reasoning.
    """

    response = llm.invoke(prompt)

    return {
        "risk_assessment": response.content
    }
```

```python
def credit_validation(state: FinanceState):

    prompt = f"""
    You are a Credit Bureau Validation Agent.

    Analyze:
    1. Credit score
    2. Repayment history
    3. Loan history
    4. Defaults
    5. Creditworthiness

    Credit Score:
    {state['credit_score']}
    """

    response = llm.invoke(prompt)

    return {
        "credit_validation": response.content
    }
```

```python
def fraud_detection(state: FinanceState):

    prompt = f"""
    You are a Financial Fraud Detection Agent.

    Analyze:
    1. Identity inconsistencies
    2. Suspicious transactions
    3. Fraud probability
    4. Fake documentation indicators
    5. Risk anomalies

    Applicant Details:
    {state['applicant_details']}
    """

    response = llm.invoke(prompt)

    return {
        "fraud_detection": response.content
    }
```

```python
def kyc_verification(state: FinanceState):

    prompt = f"""
    You are a KYC Verification Agent.

    Verify:
    1. Identity validity
    2. Address consistency
    3. Regulatory compliance
    4. Verification completeness

    Applicant Details:
    {state['applicant_details']}
    """

    response = llm.invoke(prompt)

    return {
        "kyc_verification": response.content
    }
```

```python
def employment_verification(state: FinanceState):

    prompt = f"""
    You are an Employment Verification Agent.

    Verify:
    1. Employer authenticity
    2. Salary consistency
    3. Employment duration
    4. Job stability

    Employment Type:
    {state['employment_type']}
    """

    response = llm.invoke(prompt)

    return {
        "employment_verification": response.content
    }
```

```python
def debt_to_income_analysis(state: FinanceState):

    prompt = f"""
    You are a Debt-to-Income Analysis Agent.

    Calculate:
    1. Debt-to-income ratio
    2. EMI burden
    3. Affordability score
    4. Repayment capacity

    Loan Amount:
    {state['loan_amount']}

    Financial Behavior:
    {state['financial_behavior']}
    """

    response = llm.invoke(prompt)

    return {
        "debt_to_income_ratio": response.content
    }
```

```python
def route_application(state: FinanceState):

    fraud = state["fraud_detection"].lower()
    missing = state["missing_documents"].lower()
    dti = state["debt_to_income_ratio"].lower()
    risk = state["risk_assessment"].lower()

    if "high fraud" in fraud:

        return {
            "next_route": "fraud"
        }

    elif "missing" in missing:

        return {
            "next_route": "clarification"
        }

    elif "low risk" in risk or "excellent" in risk:

        return {
            "next_route": "fasttrack"
        }

    elif "high" in dti:

        return {
            "next_route": "rejection"
        }

    else:

        return {
            "next_route": "manual"
        }


def route_selector(state: FinanceState):

    return state["next_route"]
```

```python
def fraud_agent(state: FinanceState):

    return {
        "assigned_path": "Fraud Investigation Team"
    }



def fasttrack_agent(state: FinanceState):

    return {
        "assigned_path": "Fast Track Approval"
    }

def clarification_agent(state: FinanceState):

    return {
        "assigned_path": "Document Clarification Team"
    }

def manual_review_agent(state: FinanceState):

    return {
        "assigned_path": "Manual Underwriting Review"
    }

def rejection_agent(state: FinanceState):

    return {
        "assigned_path": "Rejection Recommendation Team"
    }
```

```python
def generate_risk_summary(state: FinanceState):

    prompt = f"""
    Generate financial risk summary.

    Applicant:
    {state['applicant_profile']}

    Risk:
    {state['risk_assessment']}

    Fraud Analysis:
    {state['fraud_detection']}
    """

    response = llm.invoke(prompt)

    return {
        "risk_summary": response.content
    }

def generate_approval_rationale(state: FinanceState):

    prompt = f"""
    Generate approval or rejection rationale.

    Assigned Path:
    {state['assigned_path']}

    Risk:
    {state['risk_assessment']}

    DTI:
    {state['debt_to_income_ratio']}
    """

    response = llm.invoke(prompt)

    return {
        "approval_rationale": response.content
    }
```

```python
def recommend_loan_amount(state: FinanceState):

    prompt = f"""
    Recommend suitable loan amount.

    Applicant:
    {state['applicant_profile']}

    Income Stability:
    {state['income_stability']}

    Risk:
    {state['risk_assessment']}
    """

    response = llm.invoke(prompt)

    return {
        "loan_recommendation": response.content
    }
```

```python
def recommend_interest_slab(state: FinanceState):

    prompt = f"""
    Recommend loan interest slab.

    Credit Validation:
    {state['credit_validation']}

    Risk:
    {state['risk_assessment']}

    Financial Behavior:
    {state['financial_behavior']}
    """

    response = llm.invoke(prompt)

    return {
        "interest_slab": response.content
    }
```

```python
def generate_applicant_notification(state: FinanceState):

    prompt = f"""
    Generate professional applicant notification.

    Assigned Path:
    {state['assigned_path']}

    Loan Recommendation:
    {state['loan_recommendation']}
    """

    response = llm.invoke(prompt)

    return {
        "applicant_notification": response.content
    }
```

```python
def generate_final_report(state: FinanceState):

    prompt = f"""
    Generate comprehensive final loan decision report.

    Include:
    1. Applicant profile
    2. Risk assessment
    3. Fraud analysis
    4. Verification summary
    5. Loan recommendation
    6. Interest recommendation
    7. Approval/Rejection rationale

    Applicant:
    {state['applicant_profile']}

    Assigned Path:
    {state['assigned_path']}
    """

    response = llm.invoke(prompt)

    return {
        "final_decision_report": response.content
    }
```

```python
graph = StateGraph(FinanceState)

graph.add_node("profile", extract_profile)
graph.add_node("income", analyze_income_stability)
graph.add_node("documents", check_missing_documents)
graph.add_node("spending", analyze_spending_patterns)
graph.add_node("behavior", analyze_financial_behavior)
graph.add_node("risk", assess_risk)

graph.add_node("credit", credit_validation)
graph.add_node("fraud", fraud_detection)
graph.add_node("kyc", kyc_verification)
graph.add_node("employment", employment_verification)
graph.add_node("dti", debt_to_income_analysis)

graph.add_node("router", route_application)

graph.add_node("fraudagent", fraud_agent)
graph.add_node("fasttrack", fasttrack_agent)
graph.add_node("clarification", clarification_agent)
graph.add_node("manual", manual_review_agent)
graph.add_node("rejection", rejection_agent)

graph.add_node("summary", generate_risk_summary)
graph.add_node("rationale", generate_approval_rationale)
graph.add_node("loan", recommend_loan_amount)
graph.add_node("interest", recommend_interest_slab)
graph.add_node("notification", generate_applicant_notification)

graph.add_node("finalreport", generate_final_report)

graph.add_edge(START, "profile")
graph.add_edge(START, "income")
graph.add_edge(START, "documents")
graph.add_edge(START, "spending")
graph.add_edge(START, "behavior")

graph.add_edge("profile", "risk")
graph.add_edge("income", "risk")
graph.add_edge("documents", "risk")
graph.add_edge("spending", "risk")
graph.add_edge("behavior", "risk")

graph.add_edge("risk", "credit")
graph.add_edge("risk", "fraud")
graph.add_edge("risk", "kyc")
graph.add_edge("risk", "employment")
graph.add_edge("risk", "dti")

graph.add_edge("credit", "router")
graph.add_edge("fraud", "router")
graph.add_edge("kyc", "router")
graph.add_edge("employment", "router")
graph.add_edge("dti", "router")

graph.add_conditional_edges(
    "router",
    route_selector,
    {
        "fraud": "fraudagent",
        "fasttrack": "fasttrack",
        "clarification": "clarification",
        "manual": "manual",
        "rejection": "rejection"
    }
)

decision_nodes = [
    "fraudagent",
    "fasttrack",
    "clarification",
    "manual",
    "rejection"
]

for node in decision_nodes:

    graph.add_edge(node, "summary")
    graph.add_edge(node, "rationale")
    graph.add_edge(node, "loan")
    graph.add_edge(node, "interest")

graph.add_edge("loan", "notification")

graph.add_edge("summary", "finalreport")
graph.add_edge("rationale", "finalreport")
graph.add_edge("loan", "finalreport")
graph.add_edge("interest", "finalreport")

graph.add_edge("finalreport", END)

workflow = graph.compile()
```
